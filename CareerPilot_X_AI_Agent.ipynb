```python
# 🎯 CareerPilot X — Smart Opportunity Filter

BLOCKED_DOMAINS = [
    "instagram.com",
    "facebook.com",
    "reddit.com",
    "wikipedia.org",
    "youtube.com",
    "tiktok.com",
    "x.com",
    "twitter.com",
    "researchgate.net",
    "arxiv.org",
    "ieee.org",
    "springer.com",
    "medium.com"
]

STRONG_JOB_TERMS = [
    "internship",
    "intern",
    "job",
    "jobs",
    "hiring",
    "vacancy",
    "opening",
    "career",
    "apply",
    "recruitment",
    "trainee"
]

AI_TERMS = [
    "ai",
    "artificial intelligence",
    "machine learning",
    "ml",
    "deep learning",
    "nlp",
    "natural language processing",
    "generative ai",
    "data science",
    "computer vision",
    "llm"
]


def relevance_score(title, snippet, link, goal, location, experience):

    title = normalize(title)
    snippet = normalize(snippet)
    link = normalize(link)

    text = f"{title} {snippet}"

    # 🚫 Block social/media/academic sources
    for domain in BLOCKED_DOMAINS:
        if domain in link:
            return 0

    score = 20

    goal_text = normalize(goal)

    # 🎯 Career-goal matching
    goal_words = [
        word for word in goal_text.split()
        if len(word) >= 3
    ]

    for word in goal_words:
        if word in text:
            score += 10

    # 🤖 AI/ML relevance
    ai_hits = sum(
        1 for term in AI_TERMS
        if term in text
    )

    score += min(ai_hits * 8, 32)

    # 💼 Actual opportunity signal
    job_hits = sum(
        1 for term in STRONG_JOB_TERMS
        if term in text
    )

    score += min(job_hits * 7, 28)

    # 🇮🇳 Location
    location_text = normalize(location)

    if location_text and location_text in text:
        score += 10

    # 🎓 Fresher/student signal
    if (
        "student" in normalize(experience)
        or "fresher" in normalize(experience)
    ):
        fresher_terms = [
            "fresher",
            "freshers",
            "entry level",
            "graduate",
            "trainee",
            "intern"
        ]

        if any(term in text for term in fresher_terms):
            score += 12

    # ❌ Strongly penalize irrelevant content
    bad_terms = [
        "conference",
        "journal",
        "research paper",
        "webinar",
        "summit",
        "tutorial",
        "course",
        "admission",
        "exam",
        "event"
    ]

    for term in bad_terms:
        if term in text:
            score -= 20

    return safe_score(score)


print("✅ Smart Opportunity Filter installed!")
print("🚫 Social/media/academic noise blocked.")
print("🎯 AI/ML career relevance strengthened.")
print("💼 Job/internship signals strengthened.")
```

```python
# ============================================================
# 🔧 FINAL RELEVANCE FILTER FIX
# ============================================================

def relevance_score(title, snippet, link, goal, *extra_args):

    title = str(title or "").lower()
    snippet = str(snippet or "").lower()
    link = str(link or "").lower()
    goal = str(goal or "").lower()

    text = f"{title} {snippet}"

    # Block unwanted websites
    blocked_domains = [
        "instagram.com",
        "facebook.com",
        "reddit.com",
        "wikipedia.org",
        "youtube.com",
        "tiktok.com",
        "x.com",
        "twitter.com",
        "researchgate.net",
        "arxiv.org",
        "ieee.org",
        "springer.com",
        "medium.com"
    ]

    if any(domain in link for domain in blocked_domains):
        return 0

    # Block personal LinkedIn profiles
    if "linkedin.com/in/" in link:
        return 0

    if "linkedin.com/pub/" in link:
        return 0

    # Block personal/student pages
    profile_signals = [
        "profile of",
        "my profile",
        "student portfolio",
        "personal website",
        "about me",
        "resume of",
        "cv of",
        "pre-final year student",
        "final year student",
        "student profile",
        "portfolio of"
    ]

    if any(signal in text for signal in profile_signals):
        return 0

    # Block profile-style titles
    profile_patterns = [
        r"\bpre-final year\b",
        r"\bfinal year student\b",
        r"\bstudent at\b",
        r"\bprofile of\b",
        r"\bresume of\b",
        r"\bcv of\b",
        r"\babout me\b"
    ]

    if any(re.search(pattern, title) for pattern in profile_patterns):
        return 0

    # Actual job/internship signals
    job_signals = [
        "job",
        "jobs",
        "intern",
        "internship",
        "hiring",
        "vacancy",
        "vacancies",
        "career",
        "careers",
        "apply",
        "employment",
        "opening",
        "openings",
        "recruitment",
        "recruiting",
        "trainee",
        "fresher",
        "graduate role"
    ]

    if not any(signal in text for signal in job_signals):
        return 0

    # AI/ML signals
    ai_signals = [
        "artificial intelligence",
        "machine learning",
        "ai/ml",
        "ai engineer",
        "ml engineer",
        "data science",
        "deep learning",
        "nlp",
        "natural language processing",
        "generative ai",
        "genai",
        "computer vision",
        "llm",
        "large language model",
        "rag"
    ]

    ai_goal = any(
        word in goal
        for word in [
            "ai",
            "ml",
            "machine learning",
            "artificial intelligence",
            "data science",
            "nlp",
            "deep learning"
        ]
    )

    if ai_goal and not any(
        signal in text for signal in ai_signals
    ):
        return 0

    # Calculate relevance
    score = 35

    ai_hits = sum(
        1 for signal in ai_signals
        if signal in text
    )

    score += min(ai_hits * 8, 32)

    job_hits = sum(
        1 for signal in job_signals
        if signal in text
    )

    score += min(job_hits * 4, 16)

    goal_words = [
        word
        for word in re.findall(r"[a-zA-Z]+", goal)
        if len(word) > 2
    ]

    goal_hits = sum(
        1 for word in goal_words
        if word in text
    )

    score += min(goal_hits * 3, 12)

    return min(score, 100)


print("✅ FINAL relevance_score loaded")
print("✅ Accepts the existing 6 arguments")
print("✅ Profile filtering enabled")
print("✅ Job filtering enabled")
```

```python
# ============================================================
# 📄 CAREERPILOT X — RESUME INTELLIGENCE ENGINE
# ============================================================

import re
import os
import pandas as pd

# Common career skills we can detect
CAREER_SKILLS = [
    "python", "java", "c++", "c", "sql", "javascript",
    "html", "css", "react", "node.js", "git", "github",
    "machine learning", "deep learning", "artificial intelligence",
    "ai", "ml", "nlp", "tensorflow", "pytorch",
    "pandas", "numpy", "scikit-learn", "opencv",
    "generative ai", "genai", "rag", "langchain",
    "llm", "data science", "data analysis",
    "power bi", "tableau", "aws", "azure", "gcp",
    "docker", "kubernetes", "linux", "flask", "django",
    "fastapi", "mongodb", "mysql", "postgresql"
]


def extract_resume_text(file_path):
    """Extract text from PDF or TXT resume."""

    if not file_path:
        return ""

    # Gradio may provide a filepath string
    if hasattr(file_path, "name"):
        file_path = file_path.name

    file_path = str(file_path)

    try:

        # PDF
        if file_path.lower().endswith(".pdf"):

            from pypdf import PdfReader

            reader = PdfReader(file_path)

            pages = []

            for page in reader.pages:
                text = page.extract_text() or ""
                pages.append(text)

            return "\n".join(pages)

        # TXT
        elif file_path.lower().endswith(".txt"):

            with open(
                file_path,
                "r",
                encoding="utf-8",
                errors="ignore"
            ) as f:
                return f.read()

        return ""

    except Exception as e:

        print("Resume extraction error:", e)

        return ""


def extract_resume_skills(text):

    text_lower = text.lower()

    found = []

    for skill in CAREER_SKILLS:

        pattern = r"(?<!\w)" + re.escape(skill.lower()) + r"(?!\w)"

        if re.search(pattern, text_lower):

            found.append(skill)

    return sorted(set(found))


def resume_intelligence(resume_file, opportunities=None):

    text = extract_resume_text(resume_file)

    if not text.strip():

        return (
            "❌ Could not extract text from the resume.",
            [],
            0,
            [],
            "Please upload a text-based PDF or TXT resume."
        )

    skills = extract_resume_skills(text)

    # --------------------------------------------------------
    # Compare resume skills with live opportunities
    # --------------------------------------------------------

    matched_skills = set()
    missing_skills = set()

    opportunity_results = []

    if opportunities is not None:

        if isinstance(opportunities, pd.DataFrame):

            df = opportunities.copy()

        else:

            try:
                df = pd.DataFrame(opportunities)
            except:
                df = pd.DataFrame()

        if not df.empty:

            for _, row in df.iterrows():

                title = str(
                    row.get("Opportunity", "")
                )

                combined = title.lower()

                job_skills = []

                for skill in CAREER_SKILLS:

                    if skill.lower() in combined:

                        job_skills.append(skill)

                job_skill_set = set(job_skills)

                matched = job_skill_set.intersection(skills)

                missing = job_skill_set.difference(skills)

                matched_skills.update(matched)

                missing_skills.update(missing)

                score = 0

                if job_skill_set:

                    score = round(
                        len(matched) /
                        len(job_skill_set) * 100,
                        1
                    )

                else:

                    score = 50

                opportunity_results.append({
                    "Opportunity": title,
                    "Resume Match %": score,
                    "Matched Skills": ", ".join(
                        sorted(matched)
                    ),
                    "Potential Gaps": ", ".join(
                        sorted(missing)
                    )
                })

    # --------------------------------------------------------
    # Overall resume score
    # --------------------------------------------------------

    if opportunity_results:

        overall_score = round(
            sum(
                x["Resume Match %"]
                for x in opportunity_results
            ) / len(opportunity_results),
            1
        )

    else:

        overall_score = 0

    # --------------------------------------------------------
    # Recommendations
    # --------------------------------------------------------

    recommendations = []

    for skill in sorted(missing_skills):

        recommendations.append(
            f"📚 Learn or strengthen **{skill.title()}**"
        )

    if not recommendations:

        recommendations.append(
            "🎉 No major skill gaps detected from the current opportunities."
        )

    summary = f"""
## 📄 Resume Intelligence

### 🎯 Overall Resume Match
**{overall_score}%**

### 💻 Skills Detected
{", ".join(s.title() for s in skills) if skills else "No recognised technical skills detected."}

### ✅ Market-Matched Skills
{", ".join(s.title() for s in sorted(matched_skills)) if matched_skills else "No direct matches yet."}

### 📚 Potential Skill Gaps
{", ".join(s.title() for s in sorted(missing_skills)) if missing_skills else "No major gaps detected."}
"""

    return (
        summary,
        sorted(skills),
        overall_score,
        opportunity_results,
        "\n".join(recommendations)
    )


print("✅ Resume Intelligence Engine created successfully!")
```

```python
# ============================================================
# 🎯 CAREERPILOT X — OPPORTUNITY INTELLIGENCE ENGINE
# ============================================================

import pandas as pd
import re


def opportunity_intelligence(opportunities, user_skills=""):
    """
    Analyze live opportunities and rank the strongest choices.
    """

    if opportunities is None:
        return (
            "❌ No opportunities available.",
            pd.DataFrame()
        )

    if not isinstance(opportunities, pd.DataFrame):
        try:
            opportunities = pd.DataFrame(opportunities)
        except:
            return (
                "❌ Could not process opportunities.",
                pd.DataFrame()
            )

    if opportunities.empty:
        return (
            "❌ No opportunities found. Run Career Mission first.",
            pd.DataFrame()
        )

    user_skill_list = [
        x.strip().lower()
        for x in str(user_skills).split(",")
        if x.strip()
    ]

    results = []

    for _, row in opportunities.iterrows():

        title = str(
            row.get("Opportunity", "")
        )

        source = str(
            row.get("Source", "")
        )

        text = f"{title} {source}".lower()

        # ----------------------------------------------------
        # Extract recognizable skills
        # ----------------------------------------------------

        detected_skills = []

        for skill in CAREER_SKILLS:

            if skill.lower() in text:

                detected_skills.append(skill)

        detected_set = set(
            x.lower()
            for x in detected_skills
        )

        user_set = set(user_skill_list)

        matched = detected_set.intersection(user_set)

        missing = detected_set.difference(user_set)

        # ----------------------------------------------------
        # Existing scores
        # ----------------------------------------------------

        try:
            relevance = float(
                row.get("Relevance %", 0)
            )
        except:
            relevance = 0

        try:
            match = float(
                row.get("Match %", 0)
            )
        except:
            match = 0

        # ----------------------------------------------------
        # Skill alignment
        # ----------------------------------------------------

        if detected_set:

            skill_alignment = round(
                len(matched) /
                len(detected_set) * 100,
                1
            )

        else:

            skill_alignment = 50

        # ----------------------------------------------------
        # Remote / paid signals
        # ----------------------------------------------------

        remote = "Unknown"

        if any(
            x in text
            for x in [
                "remote",
                "work from home",
                "wfh"
            ]
        ):
            remote = "Likely Remote"

        elif "onsite" in text or "on-site" in text:
            remote = "Likely On-site"

        paid = "Unknown"

        if any(
            x in text
            for x in [
                "paid internship",
                "stipend",
                "salary",
                "₹",
                "inr"
            ]
        ):
            paid = "Potentially Paid"

        # ----------------------------------------------------
        # Overall opportunity score
        # ----------------------------------------------------

        overall = round(
            (
                relevance * 0.35
                + match * 0.35
                + skill_alignment * 0.30
            ),
            1
        )

        # ----------------------------------------------------
        # Priority
        # ----------------------------------------------------

        if overall >= 80:
            priority = "🔥 HIGH"

        elif overall >= 65:
            priority = "🟡 MEDIUM"

        else:
            priority = "⚪ LOW"

        results.append({
            "Opportunity": title,
            "Source": source,
            "Overall Score": overall,
            "Relevance": relevance,
            "Match": match,
            "Skill Alignment": skill_alignment,
            "Matched Skills": ", ".join(
                sorted(matched)
            ),
            "Potential Gaps": ", ".join(
                sorted(missing)
            ),
            "Remote Signal": remote,
            "Paid Signal": paid,
            "Priority": priority
        })

    result_df = pd.DataFrame(results)

    result_df = result_df.sort_values(
        "Overall Score",
        ascending=False
    ).reset_index(drop=True)

    # --------------------------------------------------------
    # Best opportunity
    # --------------------------------------------------------

    best = result_df.iloc[0]

    summary = f"""
## 🏆 Opportunity Intelligence

### 🎯 Best Opportunity

**{best["Opportunity"]}**

### ⭐ CareerPilot Score
**{best["Overall Score"]}%**

### 🔎 Relevance
**{best["Relevance"]}%**

### 🧠 Profile Match
**{best["Match"]}%**

### 💻 Skill Alignment
**{best["Skill Alignment"]}%**

### 🌐 Remote Signal
**{best["Remote Signal"]}**

### 💰 Payment Signal
**{best["Paid Signal"]}**

### ⚡ Priority
**{best["Priority"]}**

### ✅ Matched Skills

{best["Matched Skills"] or "No direct skill matches detected."}

### 📚 Potential Skill Gaps

{best["Potential Gaps"] or "No major gaps detected."}

---

### 🧠 Why CareerPilot recommends it

CareerPilot combines:

**Relevance + Profile Match + Skill Alignment**

to rank opportunities instead of simply returning search results.
"""

    return summary, result_df


print("✅ Opportunity Intelligence Engine created!")
```

```python
# ============================================================
# 🤖 CAREERPILOT X — INTERVIEW SIMULATOR ENGINE
# ============================================================

import re
import random


INTERVIEW_QUESTION_BANK = {

    "python": [
        "What is the difference between a list, tuple, set, and dictionary in Python?",
        "What are Python decorators and where would you use them?",
        "Explain exception handling in Python.",
        "What is the difference between == and is in Python?"
    ],

    "machine learning": [
        "What is the difference between supervised and unsupervised learning?",
        "What is overfitting and how can you prevent it?",
        "Explain the difference between classification and regression.",
        "What is cross-validation and why is it useful?"
    ],

    "artificial intelligence": [
        "What is Artificial Intelligence and how is it different from Machine Learning?",
        "What are the major applications of Artificial Intelligence?",
        "What is the difference between narrow AI and general AI?",
        "How would you evaluate an AI system?"
    ],

    "nlp": [
        "What is Natural Language Processing?",
        "What is tokenization in NLP?",
        "What is the difference between stemming and lemmatization?",
        "How would you build a basic text classification system?"
    ],

    "generative ai": [
        "What is Generative AI?",
        "What is an LLM?",
        "What is prompt engineering?",
        "What are hallucinations in Generative AI and how can they be reduced?"
    ],

    "rag": [
        "What is Retrieval-Augmented Generation?",
        "Why is RAG useful for reducing hallucinations?",
        "What are embeddings in a RAG system?",
        "Explain the basic architecture of a RAG application."
    ],

    "sql": [
        "What is the difference between WHERE and HAVING in SQL?",
        "Explain INNER JOIN and LEFT JOIN.",
        "What is database normalization?",
        "How would you find duplicate records in a SQL table?"
    ],

    "java": [
        "What are the main principles of Object-Oriented Programming?",
        "What is the difference between an interface and an abstract class in Java?",
        "Explain inheritance and polymorphism.",
        "What is exception handling in Java?"
    ]
}


def generate_interview_questions(
    career_goal,
    skills,
    number_of_questions=5
):

    skill_text = str(skills).lower()

    selected = []

    # Match questions to user skills
    for skill, questions in INTERVIEW_QUESTION_BANK.items():

        if skill in skill_text or skill in str(career_goal).lower():

            selected.extend(questions)

    # Generic questions if not enough matches
    generic = [
        "Tell me about yourself and your technical background.",
        "Describe an AI or software project you have worked on.",
        "What was the biggest technical challenge in your project?",
        "How do you debug a program when it produces an unexpected result?",
        "Why are you interested in this career?",
        "How do you keep yourself updated with new technologies?"
    ]

    selected.extend(generic)

    # Remove duplicates
    selected = list(dict.fromkeys(selected))

    random.shuffle(selected)

    return selected[:number_of_questions]


def evaluate_interview_answer(question, answer):

    answer = str(answer).strip()

    if not answer:

        return (
            "❌ No answer provided.",
            0,
            "Please provide an answer first."
        )

    words = answer.split()

    word_count = len(words)

    # Basic structured evaluation
    if word_count < 10:

        score = 35
        feedback = (
            "Your answer is too short. "
            "Add an explanation and a practical example."
        )

    elif word_count < 25:

        score = 60
        feedback = (
            "Good start. Improve the answer with "
            "more technical detail and an example."
        )

    elif word_count < 50:

        score = 80
        feedback = (
            "Good answer. You explained the concept "
            "with reasonable detail."
        )

    else:

        score = 90
        feedback = (
            "Strong answer. It contains enough detail. "
            "Try to keep it structured and concise in a real interview."
        )

    # Technical keywords
    technical_terms = [
        "because",
        "example",
        "project",
        "algorithm",
        "model",
        "data",
        "python",
        "machine learning",
        "ai",
        "nlp",
        "training",
        "testing"
    ]

    found_terms = [
        term
        for term in technical_terms
        if term in answer.lower()
    ]

    if len(found_terms) >= 3:
        score = min(100, score + 5)

    result = f"""
## 🤖 Interview Evaluation

### ⭐ Score
**{score}/100**

### 💬 Feedback
{feedback}

### 🧠 Technical signals detected
{", ".join(found_terms) if found_terms else "Add more technical terminology."}

### 📌 Improvement tip
Use this structure:

**Concept → Explanation → Example → Result**
"""

    return result, score, feedback


def interview_session(
    career_goal,
    skills,
    number_of_questions
):

    questions = generate_interview_questions(
        career_goal,
        skills,
        int(number_of_questions)
    )

    formatted = ""

    for i, question in enumerate(
        questions,
        start=1
    ):

        formatted += (
            f"### Question {i}\n"
            f"**{question}**\n\n"
        )

    return formatted


print("✅ Interview Simulator Engine created!")
```

```python
# ============================================================
# 🔮 CAREERPILOT X — WHAT-IF CAREER LAB
# ============================================================

import pandas as pd

def what_if_career_lab(
    career_goal,
    current_skills,
    additional_skills,
    target_role=None
):
    try:
        career_goal = str(career_goal).strip()
        current_skills = str(current_skills).strip()
        additional_skills = str(additional_skills).strip()

        # Convert skills into clean sets
        current = {
            x.strip().lower()
            for x in current_skills.split(",")
            if x.strip()
        }

        additional = {
            x.strip().lower()
            for x in additional_skills.split(",")
            if x.strip()
        }

        combined = current.union(additional)

        if target_role:
            target = str(target_role).strip()
        else:
            target = career_goal

        # ----------------------------------------------------
        # Skill importance by career
        # ----------------------------------------------------
        role_skills = {
            "ai/ml": [
                "python", "machine learning", "deep learning",
                "numpy", "pandas", "nlp", "tensorflow",
                "pytorch", "generative ai", "rag"
            ],
            "data science": [
                "python", "sql", "pandas", "numpy",
                "statistics", "machine learning",
                "data analysis", "power bi"
            ],
            "data analyst": [
                "sql", "excel", "python", "pandas",
                "power bi", "tableau", "data analysis"
            ],
            "software developer": [
                "python", "java", "c++", "sql",
                "git", "github", "data structures",
                "algorithms"
            ],
            "cybersecurity": [
                "python", "linux", "networking",
                "cybersecurity", "sql", "ethical hacking"
            ],
            "cloud": [
                "python", "linux", "aws",
                "azure", "docker", "kubernetes"
            ]
        }

        role_key = target.lower()

        selected_skills = []

        for role, skills in role_skills.items():
            if role in role_key or role_key in role:
                selected_skills = skills
                break

        if not selected_skills:
            selected_skills = [
                "python",
                "machine learning",
                "sql",
                "git",
                "generative ai",
                "rag"
            ]

        # ----------------------------------------------------
        # Current vs future match
        # ----------------------------------------------------
        current_match = set(selected_skills).intersection(current)
        future_match = set(selected_skills).intersection(combined)

        current_score = round(
            len(current_match) / len(selected_skills) * 100,
            1
        )

        future_score = round(
            len(future_match) / len(selected_skills) * 100,
            1
        )

        improvement = round(
            future_score - current_score,
            1
        )

        # ----------------------------------------------------
        # Newly valuable skills
        # ----------------------------------------------------
        newly_added = sorted(
            set(additional).intersection(selected_skills)
        )

        remaining_gaps = sorted(
            set(selected_skills).difference(combined)
        )

        # ----------------------------------------------------
        # Impact level
        # ----------------------------------------------------
        if improvement >= 30:
            impact = "🚀 VERY HIGH IMPACT"
        elif improvement >= 15:
            impact = "🔥 HIGH IMPACT"
        elif improvement > 0:
            impact = "🟡 MODERATE IMPACT"
        else:
            impact = "⚪ LOW IMPACT"

        # ----------------------------------------------------
        # Recommendation
        # ----------------------------------------------------
        if remaining_gaps:
            next_steps = "\n".join(
                f"{i+1}. 📚 Learn **{skill.title()}**"
                for i, skill in enumerate(remaining_gaps[:5])
            )
        else:
            next_steps = "🎉 No major skill gaps detected for this target."

        # ----------------------------------------------------
        # Skill comparison table
        # ----------------------------------------------------
        rows = []

        for skill in selected_skills:
            if skill in current:
                status = "✅ Already have"
            elif skill in additional:
                status = "🚀 Added in What-If"
            else:
                status = "📚 Missing"

            rows.append({
                "Skill": skill.title(),
                "Current Profile": "✅" if skill in current else "❌",
                "After What-If": "✅" if skill in combined else "❌",
                "Status": status
            })

        comparison_df = pd.DataFrame(rows)

        # ----------------------------------------------------
        # Final report
        # ----------------------------------------------------
        report = f"""
## 🔮 What-If Career Lab

### 🎯 Target Career
**{target}**

### 🧪 Scenario
Current Skills:

**{current_skills or "None provided"}**

Additional Skills:

**{additional_skills or "None provided"}**

---

## 📊 Career Match Simulation

| Metric | Current | After What-If |
|---|---:|---:|
| Career Match | **{current_score}%** | **{future_score}%** |
| Improvement | — | **+{improvement}%** |

### {impact}

---

## 🚀 Skills That Created the Improvement

{", ".join(x.title() for x in newly_added)
if newly_added else "No major target-role skills were added."}

---

## 📚 Remaining Skill Gaps

{", ".join(x.title() for x in remaining_gaps)
if remaining_gaps else "No major gaps detected."}

---

## 🧭 Recommended Next Steps

{next_steps}

---

### 🧠 CareerPilot Reasoning

CareerPilot simulated the target career using the selected
skill requirements and compared your current profile against
the hypothetical upgraded profile.

**Current → Scenario → Match Improvement → Remaining Gaps**
"""

        return report, comparison_df

    except Exception as e:
        return (
            f"❌ What-If Lab Error: {type(e).__name__}: {str(e)}",
            pd.DataFrame()
        )

print("✅ What-If Career Lab created successfully!")
```

```python
# ============================================================
# 🧬 CAREERPILOT X — OPPORTUNITY DNA
# ============================================================

import pandas as pd

def opportunity_dna(opportunities, index=0):

    try:
        # Convert to DataFrame if needed
        if opportunities is None:
            return "❌ No opportunities available.", pd.DataFrame()

        if not isinstance(opportunities, pd.DataFrame):
            opportunities = pd.DataFrame(opportunities)

        if opportunities.empty:
            return "❌ No opportunities found. Run Career Mission first.", pd.DataFrame()

        # Keep index safe
        index = int(index)
        index = max(0, min(index, len(opportunities) - 1))

        row = opportunities.iloc[index]

        # Basic information
        title = str(row.get("Opportunity", "Unknown Opportunity"))
        source = str(row.get("Source", "Unknown"))

        # Scores
        try:
            relevance = float(row.get("Relevance %", 0))
        except:
            relevance = 0

        try:
            match = float(row.get("Match %", 0))
        except:
            match = 0

        try:
            confidence = float(row.get("Confidence %", 0))
        except:
            confidence = 0

        # Signals
        remote = str(row.get("Remote", row.get("Remote Signal", "Unknown")))
        risk = str(row.get("Risk", "Unknown"))

        # ----------------------------------------------------
        # Skill analysis
        # ----------------------------------------------------

        title_text = title.lower()

        known_skills = [
            "python",
            "java",
            "c++",
            "sql",
            "machine learning",
            "deep learning",
            "artificial intelligence",
            "ai",
            "ml",
            "nlp",
            "tensorflow",
            "pytorch",
            "pandas",
            "numpy",
            "generative ai",
            "genai",
            "rag",
            "langchain",
            "llm",
            "data science",
            "data analysis",
            "power bi",
            "tableau",
            "aws",
            "azure",
            "docker",
            "kubernetes"
        ]

        detected_skills = [
            skill for skill in known_skills
            if skill in title_text
        ]

        # Remove duplicates
        detected_skills = list(dict.fromkeys(detected_skills))

        # ----------------------------------------------------
        # DNA components
        # ----------------------------------------------------

        components = [
            {
                "DNA Component": "🎯 Relevance",
                "Score": relevance,
                "Meaning": "How closely the opportunity matches the career search."
            },
            {
                "DNA Component": "🧠 Profile Match",
                "Score": match,
                "Meaning": "How well the opportunity matches the user's profile."
            },
            {
                "DNA Component": "🔎 Source Confidence",
                "Score": confidence,
                "Meaning": "Confidence signal based on the source and available data."
            }
        ]

        # Skill signal
        skill_score = min(100, len(detected_skills) * 20)

        components.append({
            "DNA Component": "💻 Skill Signal",
            "Score": skill_score,
            "Meaning": "Technical skills detected directly from the opportunity."
        })

        # Overall DNA score
        dna_score = round(
            relevance * 0.35 +
            match * 0.35 +
            confidence * 0.20 +
            skill_score * 0.10,
            1
        )

        # Priority
        if dna_score >= 80:
            priority = "🔥 HIGH"
        elif dna_score >= 60:
            priority = "🟡 MEDIUM"
        else:
            priority = "⚪ LOW"

        # ----------------------------------------------------
        # Strengths
        # ----------------------------------------------------

        strengths = []

        if relevance >= 75:
            strengths.append("Strong career relevance")

        if match >= 70:
            strengths.append("Strong profile match")

        if confidence >= 80:
            strengths.append("Strong source confidence")

        if detected_skills:
            strengths.append(
                "Relevant skills detected: " +
                ", ".join(x.title() for x in detected_skills)
            )

        if "remote" in remote.lower():
            strengths.append("Remote-work signal detected")

        if not strengths:
            strengths.append("No major strength signals detected")

        # ----------------------------------------------------
        # Gaps / concerns
        # ----------------------------------------------------

        concerns = []

        if relevance < 60:
            concerns.append("Low career relevance")

        if match < 60:
            concerns.append("Profile match could be stronger")

        if confidence < 70:
            concerns.append("Limited source confidence")

        if remote.lower() in ["unknown", "unknown signal"]:
            concerns.append("Remote status is unclear")

        if not detected_skills:
            concerns.append("Few technical skills detected from the title")

        if not concerns:
            concerns.append("No major warning signals detected")

        # ----------------------------------------------------
        # Explanation
        # ----------------------------------------------------

        report = f"""
## 🧬 Opportunity DNA

### 🏆 Opportunity

**{title}**

**Source:** {source}

---

## 🧬 CareerPilot DNA Score

# **{dna_score}/100**

### Priority
**{priority}**

---

## 🔬 DNA Breakdown

"""

        for component in components:
            report += (
                f"### {component['DNA Component']}\n"
                f"**{component['Score']:.1f}%**\n\n"
                f"{component['Meaning']}\n\n"
            )

        report += f"""
---

## 💪 Strength Signals

"""

        for item in strengths:
            report += f"- ✅ {item}\n"

        report += f"""

---

## ⚠️ Gap / Risk Signals

"""

        for item in concerns:
            report += f"- ⚠️ {item}\n"

        report += f"""

---

## 🌐 Opportunity Signals

**Remote:** {remote}

**Risk Signal:** {risk}

**Source Confidence:** {confidence:.1f}%

---

## 💻 Skills Detected

{", ".join(x.title() for x in detected_skills)
if detected_skills else "No technical skills detected from the opportunity title."}

---

## 🧠 CareerPilot Decision

"""

        if dna_score >= 80:
            decision = "🔥 **HIGH PRIORITY** — This opportunity deserves strong consideration."
        elif dna_score >= 60:
            decision = "🟡 **MEDIUM PRIORITY** — Consider applying after reviewing the details."
        else:
            decision = "⚪ **LOW PRIORITY** — Consider other opportunities first."

        report += decision

        # ----------------------------------------------------
        # Comparison table
        # ----------------------------------------------------

        dna_table = pd.DataFrame(components)

        return report, dna_table

    except Exception as e:

        return (
            f"❌ Opportunity DNA Error: {type(e).__name__}: {str(e)}",
            pd.DataFrame()
        )


print("✅ Opportunity DNA created successfully!")
```

```python
# ============================================================
# 🧬 CAREERPILOT X — SMART SKILL DETECTOR
# ============================================================

import re

SKILL_ALIASES = {
    "Python": ["python"],
    "Java": ["java"],
    "C++": ["c++", "cpp"],
    "SQL": ["sql", "mysql", "postgresql", "postgres"],

    "Artificial Intelligence": [
        "artificial intelligence",
        "ai"
    ],

    "Machine Learning": [
        "machine learning",
        "ml"
    ],

    "Deep Learning": [
        "deep learning"
    ],

    "NLP": [
        "nlp",
        "natural language processing"
    ],

    "Generative AI": [
        "generative ai",
        "genai",
        "gen ai"
    ],

    "RAG": [
        "rag",
        "retrieval augmented generation",
        "retrieval-augmented generation"
    ],

    "LLM": [
        "llm",
        "large language model",
        "large language models"
    ],

    "TensorFlow": ["tensorflow"],
    "PyTorch": ["pytorch"],
    "Pandas": ["pandas"],
    "NumPy": ["numpy"],

    "Scikit-learn": [
        "scikit-learn",
        "sklearn"
    ],

    "LangChain": ["langchain"],

    "Computer Vision": [
        "computer vision",
        "opencv"
    ],

    "Data Science": ["data science"],
    "Data Analysis": [
        "data analysis",
        "data analytics"
    ],

    "Power BI": ["power bi"],
    "Tableau": ["tableau"],
    "Git": ["git", "github"],
    "AWS": ["aws", "amazon web services"],
    "Azure": ["azure", "microsoft azure"],
    "Docker": ["docker"],
    "Kubernetes": ["kubernetes", "k8s"],
    "Linux": ["linux"]
}


def detect_opportunity_skills(text):

    text = str(text).lower()

    found = []

    for skill, aliases in SKILL_ALIASES.items():

        for alias in aliases:

            pattern = r"(?<!\w)" + re.escape(alias) + r"(?!\w)"

            if re.search(pattern, text):
                found.append(skill)
                break

    return sorted(set(found))


# ============================================================
# 🧪 AUTOMATIC TEST
# ============================================================

test_text = """
AI/ML Intern required skills:
Python, machine learning, deep learning,
NLP, Generative AI, RAG, SQL, TensorFlow and PyTorch.
"""

skills_found = detect_opportunity_skills(test_text)

print("✅ Smart Skill Detector is working!")
print("\n🔎 Skills detected:")

for skill in skills_found:
    print("  ✅", skill)

print("\nTotal skills detected:", len(skills_found))
```

```python
# ============================================================
# 🧬 CAREERPILOT X — SMART OPPORTUNITY DNA
# ============================================================

import pandas as pd

def opportunity_dna(opportunities, index=0):

    # Check data
    if opportunities is None:
        return "❌ No opportunities available.", pd.DataFrame()

    if not isinstance(opportunities, pd.DataFrame):
        try:
            opportunities = pd.DataFrame(opportunities)
        except:
            return "❌ Could not process opportunities.", pd.DataFrame()

    if opportunities.empty:
        return "❌ No opportunities found. Run Career Mission first.", pd.DataFrame()

    # Safe index
    index = int(index)

    if index < 0:
        index = 0

    if index >= len(opportunities):
        index = 0

    row = opportunities.iloc[index]

    # --------------------------------------------------------
    # Read ALL available opportunity information
    # --------------------------------------------------------

    text_parts = []

    for column in opportunities.columns:
        value = row.get(column, "")
        if pd.notna(value):
            text_parts.append(str(value))

    opportunity_text = " ".join(text_parts)

    # --------------------------------------------------------
    # Smart skill detection
    # --------------------------------------------------------

    detected_skills = detect_opportunity_skills(opportunity_text)

    # --------------------------------------------------------
    # Read scores
    # --------------------------------------------------------

    def safe_float(value, default=0):
        try:
            return float(value)
        except:
            return default

    relevance = safe_float(row.get("Relevance %", 0))
    match = safe_float(row.get("Match %", 0))
    confidence = safe_float(row.get("Confidence %", 0))

    # --------------------------------------------------------
    # Skill signal
    # --------------------------------------------------------

    if len(detected_skills) == 0:
        skill_signal = 0
    elif len(detected_skills) <= 2:
        skill_signal = 40
    elif len(detected_skills) <= 4:
        skill_signal = 60
    elif len(detected_skills) <= 6:
        skill_signal = 75
    elif len(detected_skills) <= 8:
        skill_signal = 90
    else:
        skill_signal = 100

    # --------------------------------------------------------
    # Remote signal
    # --------------------------------------------------------

    lower_text = opportunity_text.lower()

    if any(x in lower_text for x in [
        "remote",
        "work from home",
        "wfh",
        "remote internship"
    ]):
        remote_signal = "🌐 Likely Remote"

    elif any(x in lower_text for x in [
        "on-site",
        "onsite",
        "office"
    ]):
        remote_signal = "🏢 Likely On-site"

    else:
        remote_signal = "❓ On-site/Unknown"

    # --------------------------------------------------------
    # Risk signal
    # --------------------------------------------------------

    if any(x in lower_text for x in [
        "pay to apply",
        "registration fee",
        "joining fee",
        "deposit required"
    ]):
        risk_signal = "🔴 Potential warning"

    elif any(x in lower_text for x in [
        "internship",
        "career",
        "jobs",
        "hiring"
    ]):
        risk_signal = "🟢 Low signal"

    else:
        risk_signal = "🟡 Review manually"

    # --------------------------------------------------------
    # DNA SCORE
    # --------------------------------------------------------

    dna_score = round(
        relevance * 0.30 +
        match * 0.30 +
        confidence * 0.20 +
        skill_signal * 0.20,
        1
    )

    if dna_score >= 80:
        priority = "🔥 HIGH"
    elif dna_score >= 65:
        priority = "🟡 MEDIUM"
    else:
        priority = "⚪ LOW"

    # --------------------------------------------------------
    # Skills display
    # --------------------------------------------------------

    skills_text = (
        ", ".join(detected_skills)
        if detected_skills
        else "No recognised technical skills detected."
    )

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------

    title = str(row.get("Opportunity", "Unknown Opportunity"))
    source = str(row.get("Source", "Unknown"))

    report = f"""
## 🧬 CareerPilot Opportunity DNA

### 🎯 Opportunity
**{title}**

### 🌐 Source
**{source}**

---

### ⭐ DNA Score
# **{dna_score}/100**

### 🚦 Priority
**{priority}**

---

### 🔎 Relevance
**{relevance}%**

### 🧠 Profile Match
**{match}%**

### 🛡️ Source Confidence
**{confidence}%**

### 💻 Skill Signal
**{skill_signal}%**

---

### 💪 Strengths

- Strong opportunity relevance
- Profile compatibility: **{match}%**
- Source confidence: **{confidence}%**
- **{len(detected_skills)} technical skill signals detected**

### 🧠 Skills Detected

**{skills_text}**

### 🌐 Work Mode

{remote_signal}

### ⚠️ Risk Signal

{risk_signal}

---

### 🤖 CareerPilot Decision

**{priority} PRIORITY**

CareerPilot combines:

**Relevance + Profile Match + Source Confidence + Skill Signals**

to create the Opportunity DNA score.
"""

    # --------------------------------------------------------
    # DNA TABLE
    # --------------------------------------------------------

    dna_table = pd.DataFrame([{
        "Opportunity": title,
        "DNA Score": dna_score,
        "Priority": priority,
        "Relevance": relevance,
        "Profile Match": match,
        "Confidence": confidence,
        "Skill Signal": skill_signal,
        "Skills Detected": ", ".join(detected_skills),
        "Remote": remote_signal,
        "Risk Signal": risk_signal
    }])

    return report, dna_table


print("✅ Smart Opportunity DNA created!")
```

```python
# ============================================================
# 📈 CAREERPILOT X — CAREER MARKET INTELLIGENCE
# ============================================================

import pandas as pd
import re
from collections import Counter

def career_market_intelligence(opportunities):

    if opportunities is None:
        return "❌ No opportunities available.", pd.DataFrame()

    if not isinstance(opportunities, pd.DataFrame):
        try:
            opportunities = pd.DataFrame(opportunities)
        except:
            return "❌ Could not process opportunities.", pd.DataFrame()

    if opportunities.empty:
        return "❌ No opportunities found. Run Career Mission first.", pd.DataFrame()

    # --------------------------------------------------------
    # Combine all useful opportunity text
    # --------------------------------------------------------

    text_data = []

    for _, row in opportunities.iterrows():

        row_text = ""

        for column in opportunities.columns:
            value = row.get(column, "")

            if pd.notna(value):
                row_text += " " + str(value)

        text_data.append(row_text.lower())

    combined_text = " ".join(text_data)

    # --------------------------------------------------------
    # Detect market skills
    # --------------------------------------------------------

    market_skills = {}

    for skill, aliases in SKILL_ALIASES.items():

        count = 0

        for text in text_data:

            found = False

            for alias in aliases:

                pattern = r"(?<!\w)" + re.escape(alias) + r"(?!\w)"

                if re.search(pattern, text):
                    found = True
                    break

            if found:
                count += 1

        if count > 0:
            market_skills[skill] = count

    # --------------------------------------------------------
    # Sort skills by demand
    # --------------------------------------------------------

    sorted_skills = sorted(
        market_skills.items(),
        key=lambda x: x[1],
        reverse=True
    )

    # --------------------------------------------------------
    # User skills
    # --------------------------------------------------------

    user_skills = set()

    if "CAREER_MISSION_CACHE" in globals():

        # We intentionally don't assume a specific UI input here.
        # Current market intelligence is based on live opportunities.
        pass

    # --------------------------------------------------------
    # Demand table
    # --------------------------------------------------------

    demand_rows = []

    total_opportunities = len(opportunities)

    for skill, count in sorted_skills:

        percentage = round(
            (count / total_opportunities) * 100,
            1
        )

        demand_rows.append({
            "Skill": skill,
            "Opportunity Signals": count,
            "Market Demand %": percentage
        })

    demand_table = pd.DataFrame(demand_rows)

    # --------------------------------------------------------
    # Top skills
    # --------------------------------------------------------

    if sorted_skills:

        top_skills = sorted_skills[:8]

        top_skill_text = "\n".join(
            [
                f"{i+1}. **{skill}** — {count} opportunity signal(s)"
                for i, (skill, count) in enumerate(top_skills)
            ]
        )

        strongest_skill = sorted_skills[0][0]

    else:

        top_skill_text = "No recognised technical skill signals found."
        strongest_skill = "Not enough data"

    # --------------------------------------------------------
    # Career direction
    # --------------------------------------------------------

    ai_count = 0
    data_count = 0
    software_count = 0

    for text in text_data:

        if any(x in text for x in [
            "artificial intelligence",
            "machine learning",
            "ai/ml",
            "generative ai",
            "deep learning",
            "nlp"
        ]):
            ai_count += 1

        if any(x in text for x in [
            "data science",
            "data analyst",
            "data analysis",
            "sql",
            "power bi",
            "tableau"
        ]):
            data_count += 1

        if any(x in text for x in [
            "software developer",
            "software engineer",
            "java developer",
            "python developer",
            "web developer"
        ]):
            software_count += 1

    directions = {
        "AI / ML": ai_count,
        "Data": data_count,
        "Software Development": software_count
    }

    best_direction = max(
        directions,
        key=directions.get
    )

    # --------------------------------------------------------
    # Market strength
    # --------------------------------------------------------

    if total_opportunities >= 10:
        market_strength = "🔥 Strong"
    elif total_opportunities >= 5:
        market_strength = "🟢 Moderate"
    else:
        market_strength = "🟡 Limited sample"

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------

    report = f"""
## 📈 CareerPilot Market Intelligence

### 🌐 Live Market Sample

**{total_opportunities} opportunities analyzed**

### 📊 Market Strength

**{market_strength}**

---

### 🔥 Most Demanded Skills

{top_skill_text}

---

### 🏆 Strongest Market Skill

**{strongest_skill}**

---

### 🎯 Strongest Career Direction

**{best_direction}**

### Direction Signals

- 🤖 AI / ML — **{ai_count}**
- 📊 Data — **{data_count}**
- 💻 Software Development — **{software_count}**

---

### 🧠 CareerPilot Insight

The market signals suggest focusing your preparation around:

**{strongest_skill}**

CareerPilot derives these signals from the live opportunity data collected through **SerpApi**, rather than using a fixed static list.

> ⚠️ Market demand percentages represent signals in the current search sample, not the entire job market.
"""

    return report, demand_table


print("✅ Career Market Intelligence created!")
```

```python
# ============================================================
# 🗺️ CAREERPILOT X — PERSONALIZED CAREER ROADMAP
# ============================================================

def career_roadmap(
    career_goal,
    current_skills,
    market_table=None
):

    goal = str(career_goal).strip()
    skills_text = str(current_skills).lower()

    # --------------------------------------------------------
    # Normalize user skills
    # --------------------------------------------------------

    user_skills = set()

    for skill in SKILL_ALIASES.keys():

        aliases = SKILL_ALIASES[skill]

        for alias in aliases:

            if alias.lower() in skills_text:
                user_skills.add(skill)
                break

    # --------------------------------------------------------
    # Market skills
    # --------------------------------------------------------

    market_skills = []

    if isinstance(market_table, pd.DataFrame):

        if not market_table.empty and "Skill" in market_table.columns:

            market_skills = market_table["Skill"].astype(str).tolist()

    # --------------------------------------------------------
    # Identify gaps
    # --------------------------------------------------------

    gaps = [
        skill
        for skill in market_skills
        if skill not in user_skills
    ]

    # Remove duplicates
    gaps = list(dict.fromkeys(gaps))

    # --------------------------------------------------------
    # Learning priorities
    # --------------------------------------------------------

    priority_skills = gaps[:6]

    if not priority_skills:
        priority_skills = [
            "Machine Learning",
            "Generative AI",
            "RAG",
            "SQL"
        ]

    # --------------------------------------------------------
    # Roadmap stages
    # --------------------------------------------------------

    roadmap = []

    roadmap.append(
        "1️⃣ **Strengthen your current foundation**\n"
        "Practice Python, programming fundamentals and problem solving."
    )

    for i, skill in enumerate(priority_skills, start=2):

        roadmap.append(
            f"{i}️⃣ **Learn {skill}**\n"
            f"Understand the fundamentals and build small practical examples."
        )

    project_number = len(roadmap) + 1

    roadmap.append(
        f"{project_number}️⃣ **Build 2–3 portfolio projects**\n"
        "Create projects that demonstrate the skills demanded by the market."
    )

    project_number += 1

    roadmap.append(
        f"{project_number}️⃣ **Publish your projects**\n"
        "Upload clean projects to GitHub with README files, screenshots and setup instructions."
    )

    project_number += 1

    roadmap.append(
        f"{project_number}️⃣ **Prepare your resume**\n"
        "Highlight measurable project results and relevant technical skills."
    )

    project_number += 1

    roadmap.append(
        f"{project_number}️⃣ **Prepare for interviews**\n"
        "Practice technical questions, project explanations and behavioral questions."
    )

    project_number += 1

    roadmap.append(
        f"{project_number}️⃣ **Apply consistently**\n"
        "Track applications and prioritize opportunities with the highest CareerPilot scores."
    )

    # --------------------------------------------------------
    # Goal-specific recommendation
    # --------------------------------------------------------

    goal_lower = goal.lower()

    if "ai" in goal_lower or "machine learning" in goal_lower:

        specialization = (
            "🤖 **Recommended specialization:** AI / Machine Learning\n\n"
            "Focus on Python → Machine Learning → Deep Learning → "
            "NLP/Computer Vision → Generative AI → RAG/LLMs."
        )

    elif "data" in goal_lower:

        specialization = (
            "📊 **Recommended specialization:** Data\n\n"
            "Focus on Python → SQL → Statistics → Pandas/NumPy → "
            "Data Visualization → Machine Learning."
        )

    elif "software" in goal_lower or "developer" in goal_lower:

        specialization = (
            "💻 **Recommended specialization:** Software Development\n\n"
            "Focus on programming → DSA → OOP → Git/GitHub → "
            "backend/frontend → projects."
        )

    else:

        specialization = (
            "🎯 **Recommended specialization:** Build strong programming "
            "fundamentals and specialize according to the strongest market signals."
        )

    # --------------------------------------------------------
    # Report
    # --------------------------------------------------------

    report = f"""
## 🗺️ CareerPilot Personalized Roadmap

### 🎯 Career Target

**{goal}**

---

### 💻 Current Skill Profile

{", ".join(sorted(user_skills)) if user_skills else "No recognised skills detected."}

---

### 📚 Market-Based Skill Gaps

{", ".join(gaps) if gaps else "No major gaps detected from the current market sample."}

---

### 🚀 Your Roadmap

{chr(10).join(roadmap)}

---

### 🧠 CareerPilot Recommendation

{specialization}

---

### 🔄 Continuous Improvement

CareerPilot should continuously repeat:

**Search → Analyze → Identify Gaps → Learn → Build → Apply → Interview → Update Profile → Search Again**

> ⚠️ This roadmap is based on the current opportunity sample and should be treated as guidance rather than a guarantee of employment.
"""

    roadmap_table = pd.DataFrame({
        "Stage": list(range(1, len(roadmap) + 1)),
        "Action": [
            re.sub(r"\*\*", "", item.split("\n")[0])
            for item in roadmap
        ]
    })

    return report, roadmap_table


print("✅ Personalized Career Roadmap created!")
```

```python
# ============================================================
# 👤 CAREERPILOT X — CAREER DIGITAL TWIN
# ============================================================

def career_digital_twin(
    career_goal,
    education,
    skills,
    opportunities=None,
    market_table=None
):

    goal = str(career_goal).strip()
    education = str(education).strip()
    skills_text = str(skills).strip()

    # --------------------------------------------------------
    # Detect current skills
    # --------------------------------------------------------

    detected_skills = detect_opportunity_skills(skills_text)

    # Also support manually entered common skills
    if not detected_skills:
        detected_skills = [
            x.strip()
            for x in skills_text.split(",")
            if x.strip()
        ]

    # --------------------------------------------------------
    # Opportunity statistics
    # --------------------------------------------------------

    total_opportunities = 0
    average_match = 0
    average_relevance = 0

    if isinstance(opportunities, pd.DataFrame):

        if not opportunities.empty:

            total_opportunities = len(opportunities)

            if "Match %" in opportunities.columns:
                average_match = round(
                    pd.to_numeric(
                        opportunities["Match %"],
                        errors="coerce"
                    ).fillna(0).mean(),
                    1
                )

            if "Relevance %" in opportunities.columns:
                average_relevance = round(
                    pd.to_numeric(
                        opportunities["Relevance %"],
                        errors="coerce"
                    ).fillna(0).mean(),
                    1
                )

    # --------------------------------------------------------
    # Market skills
    # --------------------------------------------------------

    market_skills = []

    if isinstance(market_table, pd.DataFrame):

        if not market_table.empty and "Skill" in market_table.columns:

            market_skills = market_table["Skill"].astype(str).tolist()

    # --------------------------------------------------------
    # Skill gaps
    # --------------------------------------------------------

    normalized_user_skills = {
        x.lower()
        for x in detected_skills
    }

    skill_gaps = [
        skill
        for skill in market_skills
        if skill.lower() not in normalized_user_skills
    ]

    skill_gaps = list(dict.fromkeys(skill_gaps))

    # --------------------------------------------------------
    # Career readiness score
    # --------------------------------------------------------

    skill_score = min(
        100,
        len(detected_skills) * 10
    )

    opportunity_score = min(
        100,
        total_opportunities * 10
    )

    readiness_score = round(
        skill_score * 0.40 +
        average_match * 0.35 +
        opportunity_score * 0.25,
        1
    )

    if readiness_score >= 80:
        readiness = "🔥 Highly Ready"

    elif readiness_score >= 60:
        readiness = "🟢 Good Progress"

    elif readiness_score >= 40:
        readiness = "🟡 Developing"

    else:
        readiness = "🔴 Early Stage"

    # --------------------------------------------------------
    # Recommended focus
    # --------------------------------------------------------

    if skill_gaps:

        focus = skill_gaps[:5]

    else:

        focus = [
            "Deep Learning",
            "Generative AI",
            "RAG",
            "LLMs"
        ]

    # --------------------------------------------------------
    # Digital Twin report
    # --------------------------------------------------------

    report = f"""
## 👤 CareerPilot Digital Twin

### 🎯 Career Goal

**{goal}**

### 🎓 Education

**{education}**

---

### 💻 Current Skill Profile

{", ".join(sorted(detected_skills)) if detected_skills else "No recognised skills detected."}

---

### 📊 Career Metrics

| Metric | Value |
|---|---:|
| Career Readiness | **{readiness_score}%** |
| Profile Match | **{average_match}%** |
| Average Relevance | **{average_relevance}%** |
| Opportunities Analyzed | **{total_opportunities}** |

### 🚦 Readiness Status

**{readiness}**

---

### 📚 Recommended Skill Focus

{chr(10).join(
    [f"- 📚 **{skill}**" for skill in focus]
)}

---

### 🧠 Digital Twin Insight

CareerPilot currently understands your profile as:

**{education} → {goal}**

with **{len(detected_skills)} recognised technical skills** and
**{len(skill_gaps)} potential market skill gaps**.

---

### 🔄 Dynamic Profile Loop

**Profile → Search → Analyze → Learn → Apply → Interview → Update Profile**

Your Digital Twin can be updated whenever your skills, projects,
applications or career goals change.

> ⚠️ Career readiness is an internal CareerPilot estimate, not a prediction of employment.
"""

    # --------------------------------------------------------
    # Digital Twin table
    # --------------------------------------------------------

    twin_table = pd.DataFrame([{
        "Career Goal": goal,
        "Education": education,
        "Readiness %": readiness_score,
        "Profile Match %": average_match,
        "Average Relevance %": average_relevance,
        "Skills": ", ".join(sorted(detected_skills)),
        "Skill Gaps": ", ".join(skill_gaps),
        "Opportunities": total_opportunities,
        "Status": readiness
    }])

    return report, twin_table


print("✅ Career Digital Twin created!")
```

```python
# ============================================================
# 🔄 CAREERPILOT X — CLOSED-LOOP CAREER AGENT
# ============================================================

def career_closed_loop(
    career_goal,
    education,
    current_skills,
    opportunities=None,
    market_table=None,
    applications=None
):

    goal = str(career_goal).strip()
    education = str(education).strip()
    skills = str(current_skills).strip()

    # --------------------------------------------------------
    # Opportunity analysis
    # --------------------------------------------------------

    total_opportunities = 0
    average_match = 0
    average_relevance = 0

    if isinstance(opportunities, pd.DataFrame):

        if not opportunities.empty:

            total_opportunities = len(opportunities)

            if "Match %" in opportunities.columns:

                average_match = round(
                    pd.to_numeric(
                        opportunities["Match %"],
                        errors="coerce"
                    ).fillna(0).mean(),
                    1
                )

            if "Relevance %" in opportunities.columns:

                average_relevance = round(
                    pd.to_numeric(
                        opportunities["Relevance %"],
                        errors="coerce"
                    ).fillna(0).mean(),
                    1
                )

    # --------------------------------------------------------
    # Market skills
    # --------------------------------------------------------

    market_skills = []

    if isinstance(market_table, pd.DataFrame):

        if not market_table.empty and "Skill" in market_table.columns:

            market_skills = market_table["Skill"].astype(str).tolist()

    # --------------------------------------------------------
    # User skills
    # --------------------------------------------------------

    detected_skills = detect_opportunity_skills(skills)

    # --------------------------------------------------------
    # Skill gaps
    # --------------------------------------------------------

    user_skill_lower = {
        x.lower()
        for x in detected_skills
    }

    skill_gaps = [
        skill
        for skill in market_skills
        if skill.lower() not in user_skill_lower
    ]

    skill_gaps = list(dict.fromkeys(skill_gaps))

    # --------------------------------------------------------
    # Application statistics
    # --------------------------------------------------------

    total_applications = 0
    applied = 0
    interviews = 0
    offers = 0

    if applications is None:

        if "APPLICATION_TRACKER" in globals():
            applications = APPLICATION_TRACKER

    if applications:

        try:

            application_df = pd.DataFrame(applications)

            total_applications = len(application_df)

            if "Status" in application_df.columns:

                applied = int(
                    (application_df["Status"] == "Applied").sum()
                )

                interviews = int(
                    (application_df["Status"] == "Interview").sum()
                )

                offers = int(
                    (application_df["Status"] == "Offer").sum()
                )

        except:

            pass

    # --------------------------------------------------------
    # Career readiness
    # --------------------------------------------------------

    skill_score = min(
        100,
        len(detected_skills) * 10
    )

    opportunity_score = min(
        100,
        total_opportunities * 10
    )

    application_score = min(
        100,
        total_applications * 20
    )

    readiness = round(
        skill_score * 0.35 +
        average_match * 0.30 +
        opportunity_score * 0.15 +
        application_score * 0.20,
        1
    )

    # --------------------------------------------------------
    # Current stage
    # --------------------------------------------------------

    if offers > 0:

        current_stage = "🏆 Offer"

    elif interviews > 0:

        current_stage = "🎤 Interview"

    elif applied > 0:

        current_stage = "📤 Application"

    elif total_opportunities > 0:

        current_stage = "🎯 Opportunity Discovery"

    else:

        current_stage = "🔎 Discovery"

    # --------------------------------------------------------
    # Next best action
    # --------------------------------------------------------

    if skill_gaps:

        next_action = (
            f"📚 Strengthen **{skill_gaps[0]}** "
            "before applying to more opportunities."
        )

    elif total_applications == 0 and total_opportunities > 0:

        next_action = (
            "📋 Select the highest-ranked opportunities "
            "and start applying."
        )

    elif interviews == 0 and applied > 0:

        next_action = (
            "🎤 Prepare for technical interviews "
            "using CareerPilot Interview Simulator."
        )

    elif interviews > 0:

        next_action = (
            "🎯 Focus on interview preparation and "
            "project explanations."
        )

    else:

        next_action = (
            "🔎 Continue searching for stronger opportunities."
        )

    # --------------------------------------------------------
    # Closed-loop report
    # --------------------------------------------------------

    report = f"""
## 🔄 CareerPilot Closed-Loop Career Agent

### 🎯 Career Target

**{goal}**

### 🎓 Education

**{education}**

---

## 📊 Current Career State

| Metric | Value |
|---|---:|
| Career Readiness | **{readiness}%** |
| Opportunities | **{total_opportunities}** |
| Average Match | **{average_match}%** |
| Average Relevance | **{average_relevance}%** |
| Applications | **{total_applications}** |
| Interviews | **{interviews}** |
| Offers | **{offers}** |

### 🚦 Current Stage

**{current_stage}**

---

## 💻 Current Skills

{", ".join(sorted(detected_skills)) if detected_skills else "No recognised skills detected."}

---

## 📚 Skill Gaps

{", ".join(skill_gaps) if skill_gaps else "No immediate market skill gaps detected."}

---

## 🚀 Next Best Action

### {next_action}

---

## 🧠 CareerPilot Closed Loop

**🔎 Discover**

Find live opportunities using SerpApi.

↓

**🧠 Analyze**

Evaluate relevance, profile match, confidence and skill signals.

↓

**🎯 Match**

Rank opportunities using CareerPilot scoring.

↓

**📚 Improve**

Identify market skills and learning gaps.

↓

**🗺️ Prepare**

Generate a personalized roadmap and interview preparation.

↓

**📋 Apply**

Track applications and progress.

↓

**🎤 Interview**

Practice and evaluate interview answers.

↓

**👤 Update Digital Twin**

Update the career profile using the latest progress.

↓

**🔄 Discover Again**

Search for better opportunities using the improved profile.

---

### 🚀 System Status

**CareerPilot is designed as a continuous career-improvement loop rather than a one-time job search.**

> ⚠️ Readiness and recommendations are CareerPilot estimates based on the available data.
"""

    # --------------------------------------------------------
    # Loop metrics table
    # --------------------------------------------------------

    loop_table = pd.DataFrame([{
        "Career Goal": goal,
        "Current Stage": current_stage,
        "Readiness %": readiness,
        "Opportunities": total_opportunities,
        "Applications": total_applications,
        "Interviews": interviews,
        "Offers": offers,
        "Skill Gaps": ", ".join(skill_gaps),
        "Next Action": next_action
    }])

    return report, loop_table


print("✅ Closed-Loop Career Agent created!")
```

```python
# ============================================================
# 🚀 CAREERPILOT X — FINAL ALL-IN-ONE DASHBOARD
# ============================================================

import gradio as gr
import pandas as pd
import traceback

# ============================================================
# SHARED CACHE
# ============================================================

if "CAREER_MISSION_CACHE" not in globals():
    CAREER_MISSION_CACHE = {}

CAREER_MISSION_CACHE["opportunities"] = pd.DataFrame()

if "APPLICATION_TRACKER" not in globals():
    APPLICATION_TRACKER = []


# ============================================================
# HELPER — GET LIVE OPPORTUNITIES
# ============================================================

def get_cached_opportunities():
    opp = CAREER_MISSION_CACHE.get(
        "opportunities",
        pd.DataFrame()
    )

    if isinstance(opp, pd.DataFrame):
        return opp

    try:
        return pd.DataFrame(opp)
    except:
        return pd.DataFrame()


# ============================================================
# 🎯 CAREER MISSION
# ============================================================

def dashboard_career_mission(
    goal,
    education,
    skills,
    location,
    experience
):

    try:

        result = run_career_mission(
            goal,
            education,
            skills,
            location,
            experience
        )

        if isinstance(result, tuple) and len(result) == 6:

            opportunities = result[0]

            if isinstance(opportunities, pd.DataFrame):
                CAREER_MISSION_CACHE["opportunities"] = opportunities.copy()

            CAREER_MISSION_CACHE["profile"] = {
                "goal": goal,
                "education": education,
                "skills": skills,
                "location": location,
                "experience": experience
            }

            return result

        return (
            pd.DataFrame(),
            "❌ Career Mission returned an unexpected result.",
            "",
            "",
            "",
            ""
        )

    except Exception as e:

        traceback.print_exc()

        return (
            pd.DataFrame(),
            f"❌ Career Mission Error: `{e}`",
            "",
            "",
            "",
            ""
        )


# ============================================================
# 📄 RESUME INTELLIGENCE
# ============================================================

def dashboard_resume(resume_file):

    try:

        opportunities = get_cached_opportunities()

        return resume_intelligence(
            resume_file,
            opportunities
        )

    except Exception as e:

        return (
            f"❌ Resume Error: `{e}`",
            [],
            0,
            pd.DataFrame(),
            "Please upload a text-based PDF or TXT resume."
        )


# ============================================================
# 🏆 OPPORTUNITY INTELLIGENCE
# ============================================================

def dashboard_opportunity(user_skills):

    try:

        opportunities = get_cached_opportunities()

        if opportunities.empty:

            return (
                "⚠️ Please run **🎯 Career Mission** first.",
                pd.DataFrame()
            )

        return opportunity_intelligence(
            opportunities,
            user_skills
        )

    except Exception as e:

        return (
            f"❌ Opportunity Intelligence Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 🤖 INTERVIEW SIMULATOR
# ============================================================

def dashboard_generate_interview(
    career_goal,
    skills,
    number_of_questions
):

    try:

        return interview_session(
            career_goal,
            skills,
            int(number_of_questions)
        )

    except Exception as e:

        return f"❌ Interview Error: `{e}`"


def dashboard_evaluate_interview(
    question,
    answer
):

    try:

        return evaluate_interview_answer(
            question,
            answer
        )

    except Exception as e:

        return (
            f"❌ Evaluation Error: `{e}`",
            0,
            str(e)
        )


# ============================================================
# 🔮 WHAT-IF CAREER LAB — FIXED
# ============================================================

def dashboard_what_if(
    goal,
    current_skills,
    additional_skills,
    target_role
):

    try:

        result = what_if_career_lab(
            goal,
            current_skills,
            additional_skills,
            target_role
        )

        # Your function returns:
        # (report, table)

        if isinstance(result, tuple):

            if len(result) >= 2:

                report = result[0]
                table = result[1]

                if not isinstance(table, pd.DataFrame):

                    try:
                        table = pd.DataFrame(table)
                    except:
                        table = pd.DataFrame()

                return str(report), table

            if len(result) == 1:
                return str(result[0]), pd.DataFrame()

        return str(result), pd.DataFrame()

    except Exception as e:

        traceback.print_exc()

        return (
            f"❌ What-If Career Lab Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 🧬 OPPORTUNITY DNA
# ============================================================

def dashboard_dna(index):

    try:

        opportunities = get_cached_opportunities()

        if opportunities.empty:

            return (
                "⚠️ Please run **🎯 Career Mission** first.",
                pd.DataFrame()
            )

        index = int(index)

        if index < 0:
            index = 0

        if index >= len(opportunities):
            index = 0

        return opportunity_dna(
            opportunities,
            index
        )

    except Exception as e:

        return (
            f"❌ Opportunity DNA Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 📈 MARKET INTELLIGENCE
# ============================================================

def dashboard_market():

    try:

        opportunities = get_cached_opportunities()

        if opportunities.empty:

            return (
                "⚠️ Please run **🎯 Career Mission** first.",
                pd.DataFrame()
            )

        result = career_market_intelligence(
            opportunities
        )

        if isinstance(result, tuple) and len(result) == 2:

            CAREER_MISSION_CACHE["market_table"] = result[1]

        return result

    except Exception as e:

        return (
            f"❌ Market Intelligence Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 🗺️ CAREER ROADMAP
# ============================================================

def dashboard_roadmap(
    goal,
    skills
):

    try:

        opportunities = get_cached_opportunities()

        if opportunities.empty:

            return (
                "⚠️ Please run **🎯 Career Mission** first.",
                pd.DataFrame()
            )

        market_result = career_market_intelligence(
            opportunities
        )

        market_table = (
            market_result[1]
            if isinstance(market_result, tuple)
            and len(market_result) >= 2
            else pd.DataFrame()
        )

        return career_roadmap(
            goal,
            skills,
            market_table
        )

    except Exception as e:

        return (
            f"❌ Roadmap Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 📋 APPLICATION TRACKER
# ============================================================

def dashboard_add_application(
    company,
    opportunity,
    link,
    status,
    score,
    notes
):

    try:

        return add_application(
            company,
            opportunity,
            link,
            status,
            score,
            notes
        )[1]

    except Exception as e:

        return pd.DataFrame([
            {
                "Error": str(e)
            }
        ])


def dashboard_refresh_tracker():

    try:

        result = get_application_tracker()

        if isinstance(result, tuple):
            return result[1]

        return result

    except Exception as e:

        return pd.DataFrame([
            {
                "Error": str(e)
            }
        ])


def dashboard_clear_tracker():

    try:

        result = clear_application_tracker()

        if isinstance(result, tuple):
            return result[1]

        return result

    except Exception as e:

        return pd.DataFrame([
            {
                "Error": str(e)
            }
        ])


# ============================================================
# 👤 DIGITAL TWIN
# ============================================================

def dashboard_digital_twin(
    goal,
    education,
    skills
):

    try:

        opportunities = get_cached_opportunities()

        market_table = CAREER_MISSION_CACHE.get(
            "market_table",
            pd.DataFrame()
        )

        return career_digital_twin(
            goal,
            education,
            skills,
            opportunities,
            market_table
        )

    except Exception as e:

        return (
            f"❌ Digital Twin Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 🔄 CLOSED LOOP
# ============================================================

def dashboard_closed_loop(
    goal,
    education,
    skills
):

    try:

        opportunities = get_cached_opportunities()

        market_table = CAREER_MISSION_CACHE.get(
            "market_table",
            pd.DataFrame()
        )

        return career_closed_loop(
            goal,
            education,
            skills,
            opportunities,
            market_table,
            APPLICATION_TRACKER
        )

    except Exception as e:

        return (
            f"❌ Closed Loop Error: `{e}`",
            pd.DataFrame()
        )


# ============================================================
# 🚀 BUILD COMPLETE DASHBOARD
# ============================================================

with gr.Blocks(
    title="CareerPilot X",
    theme=gr.themes.Soft()
) as demo:

    # ========================================================
    # HEADER
    # ========================================================

    gr.Markdown("""
# 🚀 CareerPilot X

### Autonomous Career Intelligence Platform

**Live Web Intelligence • Career Matching • Skill Intelligence • Resume Analysis • Interview Preparation • Career Planning**

**Built for the SerpApi India Hackathon 2026 🇮🇳**

---

### 🔄 Discover → Analyze → Match → Improve → Prepare → Apply → Interview → Update → Discover Again

> Run **Career Mission first** to collect live opportunities. The other intelligence features use those results.
""")

    # ========================================================
    # 🎯 1. CAREER MISSION
    # ========================================================

    with gr.Tab("🎯 Career Mission"):

        gr.Markdown("""
## 🎯 Career Mission

Tell CareerPilot what career opportunity you are looking for.

CareerPilot searches the live web using SerpApi and analyzes the results.
""")

        with gr.Row():

            mission_goal = gr.Textbox(
                label="🎯 Career Goal",
                value="AI/ML internships"
            )

            mission_education = gr.Textbox(
                label="🎓 Education",
                value="BTech CSE-AIML"
            )

        with gr.Row():

            mission_skills = gr.Textbox(
                label="💻 Current Skills",
                value="Python, Java, AI, Machine Learning, NLP"
            )

            mission_location = gr.Textbox(
                label="📍 Location",
                value="India"
            )

        mission_experience = gr.Textbox(
            label="👩‍💻 Experience",
            value="Student"
        )

        mission_button = gr.Button(
            "🚀 Start Career Mission",
            variant="primary"
        )

        mission_opportunities = gr.Dataframe(
            label="🔎 Live Opportunities",
            interactive=False
        )

        mission_summary = gr.Markdown()

        mission_skills_output = gr.Markdown()

        mission_market_output = gr.Markdown()

        mission_roadmap_output = gr.Markdown()

        mission_log = gr.Markdown()

        mission_button.click(
            dashboard_career_mission,
            inputs=[
                mission_goal,
                mission_education,
                mission_skills,
                mission_location,
                mission_experience
            ],
            outputs=[
                mission_opportunities,
                mission_summary,
                mission_skills_output,
                mission_market_output,
                mission_roadmap_output,
                mission_log
            ]
        )

    # ========================================================
    # 📄 2. RESUME INTELLIGENCE
    # ========================================================

    with gr.Tab("📄 Resume Intelligence"):

        gr.Markdown("""
## 📄 Resume Intelligence

Upload your resume and compare it against the live opportunities discovered by CareerPilot.
""")

        resume_file = gr.File(
            label="📄 Upload Resume",
            file_types=[".pdf", ".txt"],
            type="filepath"
        )

        resume_button = gr.Button(
            "🔍 Analyze Resume",
            variant="primary"
        )

        resume_summary = gr.Markdown()

        resume_skills_output = gr.JSON(
            label="💻 Detected Skills"
        )

        resume_score = gr.Number(
            label="🎯 Overall Resume Match %"
        )

        resume_table = gr.Dataframe(
            label="📊 Opportunity Resume Match",
            interactive=False
        )

        resume_recommendations = gr.Markdown()

        resume_button.click(
            dashboard_resume,
            inputs=resume_file,
            outputs=[
                resume_summary,
                resume_skills_output,
                resume_score,
                resume_table,
                resume_recommendations
            ]
        )

    # ========================================================
    # 🏆 3. OPPORTUNITY INTELLIGENCE
    # ========================================================

    with gr.Tab("🏆 Opportunity Intelligence"):

        gr.Markdown("""
## 🏆 Opportunity Intelligence

Rank opportunities using:

**Relevance + Profile Match + Skill Alignment**
""")

        opportunity_skills_input = gr.Textbox(
            label="💻 Your Skills",
            value="Python, Java, AI, Machine Learning, NLP"
        )

        opportunity_button = gr.Button(
            "🏆 Analyze Opportunities",
            variant="primary"
        )

        opportunity_report = gr.Markdown()

        opportunity_table = gr.Dataframe(
            label="🏆 Ranked Opportunities",
            interactive=False
        )

        opportunity_button.click(
            dashboard_opportunity,
            inputs=opportunity_skills_input,
            outputs=[
                opportunity_report,
                opportunity_table
            ]
        )

    # ========================================================
    # 🤖 4. INTERVIEW SIMULATOR
    # ========================================================

    with gr.Tab("🤖 Interview Simulator"):

        gr.Markdown("""
## 🤖 Interview Simulator

Generate questions based on your target career and skills,
then practice and evaluate your answers.
""")

        interview_goal_input = gr.Textbox(
            label="🎯 Career Goal",
            value="AI/ML internships"
        )

        interview_skills_input = gr.Textbox(
            label="💻 Skills",
            value="Python, Machine Learning, AI, NLP"
        )

        interview_question_count = gr.Slider(
            minimum=1,
            maximum=10,
            value=5,
            step=1,
            label="Number of Questions"
        )

        generate_interview_button = gr.Button(
            "🎤 Generate Interview Questions",
            variant="primary"
        )

        interview_questions_output = gr.Markdown()

        generate_interview_button.click(
            dashboard_generate_interview,
            inputs=[
                interview_goal_input,
                interview_skills_input,
                interview_question_count
            ],
            outputs=interview_questions_output
        )

        gr.Markdown("---")

        gr.Markdown("### 📝 Practice Your Answer")

        interview_question_input = gr.Textbox(
            label="Question"
        )

        interview_answer_input = gr.Textbox(
            label="Your Answer",
            lines=6,
            placeholder="Type your answer here..."
        )

        evaluate_interview_button = gr.Button(
            "🧠 Evaluate My Answer",
            variant="primary"
        )

        interview_evaluation = gr.Markdown()

        interview_score = gr.Number(
            label="⭐ Interview Score"
        )

        interview_feedback = gr.Markdown()

        evaluate_interview_button.click(
            dashboard_evaluate_interview,
            inputs=[
                interview_question_input,
                interview_answer_input
            ],
            outputs=[
                interview_evaluation,
                interview_score,
                interview_feedback
            ]
        )

    # ========================================================
    # 🔮 5. WHAT-IF CAREER LAB
    # ========================================================

    with gr.Tab("🔮 What-If Career Lab"):

        gr.Markdown("""
## 🔮 What-If Career Lab

Test how adding new skills can change your career readiness.
""")

        whatif_goal_input = gr.Textbox(
            label="🎯 Career Goal",
            value="AI/ML internships"
        )

        whatif_current_input = gr.Textbox(
            label="💻 Current Skills",
            value="Python, Java, AI, Machine Learning, NLP"
        )

        whatif_additional_input = gr.Textbox(
            label="🚀 Additional Skills",
            value="Generative AI, RAG, SQL"
        )

        whatif_role_input = gr.Textbox(
            label="🏆 Target Role",
            value="AI/ML Engineer"
        )

        whatif_button = gr.Button(
            "🔮 Run What-If Simulation",
            variant="primary"
        )

        whatif_report = gr.Markdown(
            label="What-If Analysis"
        )

        whatif_table = gr.Dataframe(
            label="📊 Skill Impact Analysis",
            interactive=False
        )

        # IMPORTANT:
        # TWO outputs because what_if_career_lab returns
        # report + table.

        whatif_button.click(
            dashboard_what_if,
            inputs=[
                whatif_goal_input,
                whatif_current_input,
                whatif_additional_input,
                whatif_role_input
            ],
            outputs=[
                whatif_report,
                whatif_table
            ]
        )

    # ========================================================
    # 🧬 6. OPPORTUNITY DNA
    # ========================================================

    with gr.Tab("🧬 Opportunity DNA"):

        gr.Markdown("""
## 🧬 Opportunity DNA

Deep-analyze a selected opportunity using:

**Relevance + Profile Match + Source Confidence + Skill Signal**
""")

        dna_index = gr.Number(
            label="Opportunity Number",
            value=0,
            precision=0
        )

        dna_button = gr.Button(
            "🧬 Analyze Opportunity DNA",
            variant="primary"
        )

        dna_report = gr.Markdown()

        dna_table = gr.Dataframe(
            label="🧬 Opportunity DNA",
            interactive=False
        )

        dna_button.click(
            dashboard_dna,
            inputs=dna_index,
            outputs=[
                dna_report,
                dna_table
            ]
        )

    # ========================================================
    # 📈 7. MARKET INTELLIGENCE
    # ========================================================

    with gr.Tab("📈 Market Intelligence"):

        gr.Markdown("""
## 📈 Career Market Intelligence

Analyze the skills appearing across your live opportunity sample.
""")

        market_button = gr.Button(
            "📈 Analyze Career Market",
            variant="primary"
        )

        market_report = gr.Markdown()

        market_table_output = gr.Dataframe(
            label="📊 Market Skill Demand",
            interactive=False
        )

        market_button.click(
            dashboard_market,
            outputs=[
                market_report,
                market_table_output
            ]
        )

    # ========================================================
    # 🗺️ 8. CAREER ROADMAP
    # ========================================================

    with gr.Tab("🗺️ Career Roadmap"):

        gr.Markdown("""
## 🗺️ Personalized Career Roadmap

Build a step-by-step plan based on your current profile and market signals.
""")

        roadmap_goal_input = gr.Textbox(
            label="🎯 Career Goal",
            value="AI/ML internships"
        )

        roadmap_skills_input = gr.Textbox(
            label="💻 Current Skills",
            value="Python, Java, AI, Machine Learning, NLP"
        )

        roadmap_button = gr.Button(
            "🗺️ Generate Career Roadmap",
            variant="primary"
        )

        roadmap_report = gr.Markdown()

        roadmap_table_output = gr.Dataframe(
            label="🗺️ Roadmap",
            interactive=False
        )

        roadmap_button.click(
            dashboard_roadmap,
            inputs=[
                roadmap_goal_input,
                roadmap_skills_input
            ],
            outputs=[
                roadmap_report,
                roadmap_table_output
            ]
        )

    # ========================================================
    # 📋 9. APPLICATION TRACKER
    # ========================================================

    with gr.Tab("📋 Application Tracker"):

        gr.Markdown("""
## 📋 Application Tracker

Save and track your internship/job applications.
""")

        with gr.Row():

            application_company = gr.Textbox(
                label="🏢 Company"
            )

            application_opportunity = gr.Textbox(
                label="💼 Opportunity"
            )

        with gr.Row():

            application_link = gr.Textbox(
                label="🔗 Application Link"
            )

            application_status = gr.Dropdown(
                choices=[
                    "Saved",
                    "Applied",
                    "Interview",
                    "Rejected",
                    "Offer"
                ],
                value="Saved",
                label="📌 Status"
            )

        application_score = gr.Number(
            label="⭐ CareerPilot Score",
            value=0
        )

        application_notes = gr.Textbox(
            label="📝 Notes"
        )

        add_application_button = gr.Button(
            "➕ Add Application",
            variant="primary"
        )

        tracker_table = gr.Dataframe(
            label="📋 My Applications",
            interactive=False
        )

        add_application_button.click(
            dashboard_add_application,
            inputs=[
                application_company,
                application_opportunity,
                application_link,
                application_status,
                application_score,
                application_notes
            ],
            outputs=tracker_table
        )

        refresh_tracker_button = gr.Button(
            "🔄 Refresh Tracker"
        )

        refresh_tracker_button.click(
            dashboard_refresh_tracker,
            outputs=tracker_table
        )

        clear_tracker_button = gr.Button(
            "🗑️ Clear Tracker"
        )

        clear_tracker_button.click(
            dashboard_clear_tracker,
            outputs=tracker_table
        )

    # ========================================================
    # 👤 10. CAREER DIGITAL TWIN
    # ========================================================

    with gr.Tab("👤 Career Digital Twin"):

        gr.Markdown("""
## 👤 Career Digital Twin

Build a dynamic representation of your career profile,
skills, opportunities and readiness.
""")

        twin_goal_input = gr.Textbox(
            label="🎯 Career Goal",
            value="AI/ML internships"
        )

        twin_education_input = gr.Textbox(
            label="🎓 Education",
            value="BTech CSE-AIML"
        )

        twin_skills_input = gr.Textbox(
            label="💻 Skills",
            value="Python, Java, AI, Machine Learning, NLP"
        )

        twin_button = gr.Button(
            "👤 Generate Digital Twin",
            variant="primary"
        )

        twin_report = gr.Markdown()

        twin_table = gr.Dataframe(
            label="👤 Digital Twin Metrics",
            interactive=False
        )

        twin_button.click(
            dashboard_digital_twin,
            inputs=[
                twin_goal_input,
                twin_education_input,
                twin_skills_input
            ],
            outputs=[
                twin_report,
                twin_table
            ]
        )

    # ========================================================
    # 🔄 11. CLOSED LOOP
    # ========================================================

    with gr.Tab("🔄 Closed-Loop Agent"):

        gr.Markdown("""
## 🔄 Closed-Loop Career Agent

CareerPilot connects the complete career journey:

**Discover → Analyze → Match → Improve → Prepare → Apply → Interview → Update → Discover Again**
""")

        loop_goal_input = gr.Textbox(
            label="🎯 Career Goal",
            value="AI/ML internships"
        )

        loop_education_input = gr.Textbox(
            label="🎓 Education",
            value="BTech CSE-AIML"
        )

        loop_skills_input = gr.Textbox(
            label="💻 Current Skills",
            value="Python, Java, AI, Machine Learning, NLP"
        )

        loop_button = gr.Button(
            "🔄 Run Closed-Loop Analysis",
            variant="primary"
        )

        loop_report = gr.Markdown()

        loop_table = gr.Dataframe(
            label="🔄 Career Loop Metrics",
            interactive=False
        )

        loop_button.click(
            dashboard_closed_loop,
            inputs=[
                loop_goal_input,
                loop_education_input,
                loop_skills_input
            ],
            outputs=[
                loop_report,
                loop_table
            ]
        )

    # ========================================================
    # FOOTER
    # ========================================================

    gr.Markdown("""
---

# 🚀 CareerPilot X

### Autonomous Career Intelligence Platform

**Live Web Intelligence • Career Matching • Skill Intelligence • Resume Analysis • Interview Preparation • Career Planning**

**Built for the SerpApi India Hackathon 2026 🇮🇳**

> CareerPilot X transforms live web search data into personalized career decisions, actionable skill recommendations, interview preparation and career planning.

⚠️ **Disclaimer:** CareerPilot's relevance, confidence, risk and skill signals are heuristic estimates. Always verify opportunity details directly with the original source before applying.
""")


# ============================================================
# 🚀 LAUNCH
# ============================================================

print("🚀 CareerPilot X — launching complete dashboard...")

demo.launch(
    share=True,
    debug=True
)
```
